```python
df_environmental_shares = pd.read_csv("environmental_shares_flanders.csv")
df_environmental_shares
```

```python
percent_cols = [col for col in df_environmental_shares.columns if '%' in col]
for col in percent_cols:
    df_environmental_shares[col] = df_environmental_shares[col].str.replace(',', '.').astype(float)
```

```python
average_climate = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Climat']
    .mean()
)
```

```python
average_climate = average_climate.sort_values('% Climat', ascending=False)
```

```python
channel_types = {
    'radio-1' : 'Public',
    'radio-2' : 'Public',
    'qmusic' : 'Commercial',
    'stu-bru' : 'Commercial', 
    'canvas' : 'Public',
    'play' : 'Commercial',
    'vrt1' : 'Public',
    'vtm' : 'Commercial'
}

average_climate['Type'] = average_climate['Channel Name'].map(channel_types)
```

```python
fig = px.bar(
    average_climate,
    x='Channel Name',
    y='% Climat',
    color='Type',
    title='Average % Climate Coverage by Channel Name',
    labels={'% Climat': 'Average % Climate', 'Channel Name': 'Channel Name'},
    color_discrete_map={'Commercial': 'purple', 'Public': 'green'},
    category_orders={'Channel Name': average_climate['Channel Name']}
)

fig.update_layout(
    yaxis_title='Average % Climate',
    yaxis_tickformat='.1%',  
)

fig.show()
```

```python
average_env_total = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Environnement Total']
    .mean()
)
average_env_total = average_env_total.sort_values('% Environnement Total', ascending=False)
average_env_total['Type'] = average_env_total['Channel Name'].map(channel_types)

fig = px.bar(
    average_env_total,
    x='Channel Name',
    y='% Environnement Total',
    color='Type',
    title='Average % Environnement Total by Channel Name',
    labels={'% Environnement Total': 'Average % Environnement Total', 'Channel Name': 'Channel Name'},
    color_discrete_map={'Commercial': 'purple', 'Public': 'green'},
    category_orders={'Channel Name': average_env_total['Channel Name']}
)

fig.update_layout(
    yaxis_title='Average % Climate',
    yaxis_tickformat='.1%',  
)

fig.show()
```

```python
# Calculating the different averages for Climate Change

average_climate_cause = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Climat Cause']
    .mean()
)

average_climate_sol_adapt = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Climat Solutions Adaptation']
    .mean()
)

average_climate_cons = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Climat Consequences']
    .mean()
)

average_climate_sol_atten = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Climat Solutions Attenuation']
    .mean()
)
```

```python
# Calculating the different averages for Biodiversity

average_biodiversity = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Biodiversite']
    .mean()
)

average_bio_constat = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Biodiversité Constat']
    .mean()
)

average_bio_cons = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Biodiversité Conséquences']
    .mean()
)

average_bio_causes = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Biodiversité Causes']
    .mean()
)
```

```python
# Calculating the different averages for Resources

average_resources = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Ressources']
    .mean()
)

average_res_constat = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Ressources Constat']
    .mean()
)

average_res_sol = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Ressources Solutions']
    .mean()
)
```

```python
# Merge all climate averages
merged_averages_climate = average_climate_cause
merged_averages_climate = pd.merge(merged_averages_climate, average_climate_sol_adapt, on='Channel Name', how='outer')
merged_averages_climate = pd.merge(merged_averages_climate, average_climate_cons, on='Channel Name', how='outer')
merged_averages_climate = pd.merge(merged_averages_climate, average_climate_sol_atten, on='Channel Name', how='outer')
merged_averages_climate = pd.merge(merged_averages_climate, average_climate, on='Channel Name', how='outer')
```

```python
# Merge all biodiversity averages
merged_averages_bio = average_biodiversity
merged_averages_bio = pd.merge(merged_averages_bio, average_bio_constat, on='Channel Name', how='outer')
merged_averages_bio = pd.merge(merged_averages_bio, average_bio_cons, on='Channel Name', how='outer')
merged_averages_bio = pd.merge(merged_averages_bio, average_bio_causes, on='Channel Name', how='outer')
merged_averages_bio = merged_averages_bio.sort_values('% Biodiversite', ascending=False)
```

```python
# Merge all resource averages
merged_averages_resources = average_resources
merged_averages_resources = pd.merge(merged_averages_resources, average_res_sol, on='Channel Name', how='outer')
merged_averages_resources = pd.merge(merged_averages_resources, average_res_constat, on='Channel Name', how='outer')
merged_averages_resources = merged_averages_resources.sort_values('% Ressources', ascending=False)
```

```python
# Merge all major averages
merged_averages= average_resources
merged_averages = pd.merge(merged_averages, average_climate, on='Channel Name', how='outer')
merged_averages = pd.merge(merged_averages, average_biodiversity, on='Channel Name', how='outer')
merged_averages = pd.merge(merged_averages, average_env_total, on='Channel Name', how='outer')
```

```python
df_environmental_shares = df_environmental_shares.rename(columns={
    '% Environnement Total': '% Environment Total',
    '% Climat': '% Climate',
    '% Biodiversite': '% Biodiversity',
    '% Ressources': '% Resources'
})

avg_total = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Environment Total']
    .mean()
)

avg_climat = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Climate']
    .mean()
)

avg_bio = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Biodiversity']
    .mean()
)

avg_res = (
    df_environmental_shares
    .groupby('Channel Name', as_index=False)['% Resources']
    .mean()
)

merged_averages = avg_total
merged_averages = pd.merge(merged_averages, avg_climat, on='Channel Name', how='outer')
merged_averages = pd.merge(merged_averages, avg_bio, on='Channel Name', how='outer')
merged_averages = pd.merge(merged_averages, avg_res, on='Channel Name', how='outer')

merged_averages = merged_averages.sort_values('% Environment Total', ascending=False)

fig = px.bar(
    merged_averages,
    x='Channel Name',
    y=['% Environment Total', '% Climate', '% Biodiversity', '% Resources'],
    title='% of Environmental Total and subcategories by Channel',
    labels={'value': 'Average %', 'variable': 'Subcategory'},
    color_discrete_sequence=px.colors.qualitative.Pastel,
    barmode='group'
)

fig.update_layout(
    yaxis_title='Average %',
    yaxis_tickformat='.1%',
    legend_title_text='Environmental Subcategory'
)

fig.show()
fig.write_html("charts/Media_Coverage_per_Channel.html", full_html=True, include_plotlyjs='cdn')
```

```python
df_environmental_shares['Start'] = pd.to_datetime(df_environmental_shares['Start'])

df_environmental_shares = df_environmental_shares.set_index('Start')
```

```python
weekly_data = df_environmental_shares[['% Environnement Total', '% Climat', '% Biodiversite', '% Ressources']].resample('W').sum()
```

```python
pivoted_data = df_environmental_shares.pivot_table(
    index='Start',
    columns='Channel Name',
    values='% Environnement Total',
    aggfunc='sum',
    fill_value=0
)

weekly_env_total = pivoted_data.resample('W').sum().reset_index()

fig = go.Figure()


for channel in weekly_env_total.columns[1:]:  
    fig.add_trace(go.Scatter(
        x=weekly_env_total['Start'],
        y=weekly_env_total[channel],
        name=channel, 
        mode='lines+markers',
        hovertemplate='%{fullData.name}: %{y:.2f}%<extra></extra>',
        line_shape='spline'
    ))


fig.update_layout(
    height=800,
    title='Weekly Sum of % Environnement Total by Channel',
    xaxis_title='Date',
    yaxis_title='Sum of % Environnement Total',
    hovermode='x unified',
)

fig.show()
fig.write_html("charts/Environmental_Coverage_pChannel.html", full_html=True, include_plotlyjs='cdn')
```
