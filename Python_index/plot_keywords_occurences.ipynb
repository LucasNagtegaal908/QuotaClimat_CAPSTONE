```python
import pandas as pd
import plotly.express as px

keywords = pd.read_csv("thematics_keywords_new.csv")
keywords
```

```python
# Renaming the column 0 to "Count"
counts_keywords = keywords.groupby("Keyword").size().reset_index(name="Count")
counts_keywords
```

```python
# Sorting counts_keywords by count (in descending order)
counts_keywords = counts_keywords.sort_values(by="Count", ascending=False)
# Selecting the top 60 most occuring keywords as in this case all words can be read on the x-axis
top_counts = counts_keywords.head(60)

# Creating a bar chart using plotly express
fig = px.bar(top_counts, x="Keyword", y="Count", title="Keyword Frequency")
fig.update_xaxes(tickangle=90)

fig.show()
fig.write_html("charts/Keyword_count.html", full_html=True, include_plotlyjs='cdn')
```
