````python
import os
import json
import time
from enum import Enum
from typing import List, Optional
from pydantic import BaseModel, Field
from google import genai
from google.genai import types

# ---------------------------------------------------------
# GLOBAL CONFIGURATION & POLICIES
# ---------------------------------------------------------
BATCH_SIZE = 100                    # 1,000 rows / 100 = 10 requests total (Safe under your 20 RPD cap)
TEMP_CHECKPOINT_FILE = "frame_analysis_checkpoint.csv"
FINAL_OUTPUT_FILE = "climate_frames_analysis.csv"

class ClimateCategory(str, Enum):
    SCIENCE = "Science / Expertise"
    POLICY = "Policy"
    ECONOMY = "Economy"

# ---------------------------------------------------------
# 1. STRUCTURAL RESPONSE BLUEPRINTS
# ---------------------------------------------------------
class CategorizationResultItem(BaseModel):
    ID: str = Field(description="The unique identifier string of the transcript.")
    category: Optional[ClimateCategory] = Field(
        default=None,
        description="The primary climate change category based on priority rules, or null if unrelated."
    )
    reason: str = Field(description="A 1-sentence analytical reason detailing which rules were applied.")

class FrameResponseSchema(BaseModel):
    results: List[CategorizationResultItem]

# ---------------------------------------------------------
# 2. HELPER STREAM ENGINE
# ---------------------------------------------------------
def chunk_dataframe(df: pd.DataFrame, size: int):
    for i in range(0, len(df), size):
        yield df.iloc[i : i + size]

# ---------------------------------------------------------
# 3. CORE CATEGORIZATION SYSTEM WITH AUTO-SAVE
# ---------------------------------------------------------
def categorize_climate_frames_with_recovery(df: pd.DataFrame):
    client = genai.Client(api_key="...")
    model_name = "gemini-2.5-flash"
    
    processed_ids = set()
    if os.path.exists(TEMP_CHECKPOINT_FILE):
        try:
            old_df = pd.read_csv(TEMP_CHECKPOINT_FILE)
            processed_ids = set(old_df['ID'].astype(str).tolist())
            print(f"Resuming from local file. Found {len(processed_ids)} already categorized records.")
        except Exception:
            print("Checkpoint empty or unreadable, starting fresh.")

    working_df = df[~df['ID'].isin(processed_ids)].copy()
    if working_df.empty:
        print("All excerpts have already been successfully analyzed!")
        return

    batches = list(chunk_dataframe(working_df, BATCH_SIZE))
    total_batches = len(batches)
    
    print(f"Initializing audit pipeline for {len(working_df)} rows across {total_batches} total requests...")
    
    system_instruction = (
        "You are an expert media content auditor specializing in framing analysis. "
        "Your task is to classify Flemish media log snippets into exactly ONE of the following categories, using a strict priority sequence:\n\n"
        "1. 'Science / Expertise': Contains scientific data, climate statistics, or insights shared by an expert. Policy advice backed up by statistics MUST be classified here.\n"
        "2. 'Policy': Focuses on climate-related rules, regulations, agreements, or strategies (including oil/gas policy) inside or outside Belgium.\n"
        "3. 'Economy': Explores economic impacts of climate change or financial/industrial shifts in related sectors (like oil and gas developments).\n\n"
        "STRICT PRIORITY RULE: If an excerpt exhibits indicators of multiple categories, you must prioritize 'Science / Expertise' first. "
        "If that is absent, prioritize 'Policy'. Only choose 'Economy' if it is completely isolated from the other two frames. "
        "If the excerpt does not explicitly or implicitly discuss climate change or related environmental sectors, do not assign a label (return null)."
    )

    for idx, batch_df in enumerate(batches, 1):
        print(f"Transmitting Batch {idx}/{total_batches} ({len(batch_df)} rows) to Gemini...")
        
        formatted_items = []
        for _, row in batch_df.iterrows():
            formatted_items.append(
                f"ID: {row['ID']}\nProgram: {row['Channel Program']} ({row['Channel Name']})\nExcerpt: {row['Plaintext']}\n---"
            )
            
        prompt = "Analyze these records and assign the single highest-priority category:\n\n" + "\n".join(formatted_items)
        batch_results_to_save = []
        
        try:
            response = client.models.generate_content(
                model=model_name,
                contents=prompt,
                config=types.GenerateContentConfig(
                    system_instruction=system_instruction,
                    response_mime_type="application/json",
                    response_schema=FrameResponseSchema,
                    temperature=0.0,
                ),
            )
            
            clean_text = response.text.strip()
            if clean_text.startswith("```"):
                clean_text = clean_text.split("json")[-1].strip("` \n")
                
            parsed_json = json.loads(clean_text)
            batch_results_to_save = parsed_json.get("results", [])
            print(f"Batch {idx} read and classified cleanly!")
            
        except Exception as e:
            print(f"Batch {idx} hit a formatting/parsing anomaly: {e}")
            print(f"Splitting this specific batch into micro-chunks of 5 items to isolate the bad row...")
            
            micro_chunks = list(chunk_dataframe(batch_df, 5))
            for midx, micro_df in enumerate(micro_chunks, 1):
                micro_items = [f"ID: {row['ID']}\nExcerpt: {row['Plaintext']}\n---" for _, row in micro_df.iterrows()]
                micro_prompt = "Assign high-priority climate category:\n\n" + "\n".join(micro_items)
                try:
                    micro_response = client.models.generate_content(
                        model=model_name,
                        contents=micro_prompt,
                        config=types.GenerateContentConfig(
                            system_instruction=system_instruction,
                            response_mime_type="application/json",
                            response_schema=FrameResponseSchema,
                            temperature=0.0,
                        ),
                    )
                    m_text = micro_response.text.strip()
                    if m_text.startswith("```"):
                        m_text = m_text.split("json")[-1].strip("` \n")
                    batch_results_results = json.loads(m_text).get("results", [])
                    batch_results_to_save.extend(batch_results_results)
                except Exception:
                    for _, row in micro_df.iterrows():
                        batch_results_to_save.append({"ID": str(row['ID']), "category": None, "reason": "Failed parsing check"})
                time.sleep(1)

        # --- SAVE IMMEDIATELY TO HARD DRIVE ---
        if batch_results_to_save:
            current_checkpoint_df = pd.DataFrame(batch_results_to_save)
            if not os.path.exists(TEMP_CHECKPOINT_FILE):
                current_checkpoint_df.to_csv(TEMP_CHECKPOINT_FILE, index=False)
            else:
                current_checkpoint_df.to_csv(TEMP_CHECKPOINT_FILE, mode='a', header=False, index=False)
            print(f"Local progress saved! Checkpoint file appended with +{len(batch_results_to_save)} rows.")

        # --- MANDATORY TIME-OUT TO PREVENT QUOTA CEILINGS ---
        if idx < total_batches:
            print("Sleeping for 45 seconds to guarantee quota safety...")
            time.sleep(45)

# ---------------------------------------------------------
# 4. EXECUTE PIPELINE COHORT
# ---------------------------------------------------------
if __name__ == "__main__":
    # Target your specific dataframe name variable directly
    if 'sampled_excerpts_list' in locals() or 'sampled_excerpts_list' in globals():
        # Drop entries missing plain text and sample down exactly to your top 1000 items
        working_df = sampled_excerpts_list.dropna(subset=['Plaintext']).head(1000).copy()
        working_df['ID'] = working_df['ID'].astype(str)
        
        # Execute processing loop
        categorize_climate_frames_with_recovery(working_df)
        
        # --- COMPILE FINAL SEAMLESS FILE ---
        if os.path.exists(TEMP_CHECKPOINT_FILE):
            print("\Stitching frames back to target dataset...")
            results_df = pd.read_csv(TEMP_CHECKPOINT_FILE).drop_duplicates(subset=['ID'])
            results_df['ID'] = results_df['ID'].astype(str)
            
            final_dataset = pd.merge(working_df, results_df, on='ID', how='left')
            final_dataset.to_csv(FINAL_OUTPUT_FILE, index=False, encoding='utf-8')
            print(f"Mission Successful! Consolidated files saved to: '{FINAL_OUTPUT_FILE}'")
    else:
        print("Please ensure your 'sampled_excerpts_list' DataFrame is loaded into memory before running.")
````

````python
# If High Risk Of False Positive = True, delete from thematics_keywords DataFrame
thematics_keywords = thematics_keywords[thematics_keywords["High Risk Of False Positive"] != True]
thematics_keywords
````

````python
# Getting a list of all unique keywords in the Keyword column of thematics_keywords DataFrame
all_unique_keywords = set()
for keywords in thematics_keywords["Keyword"]:
    if pd.notna(keywords):
        all_unique_keywords.update([kw.strip() for kw in keywords.split(",")])
print(f"Total unique keywords (all): {len(all_unique_keywords)}")
````

````python
# Filtering to make sure all unique keywords (non-HRFP) are selected as target keywords, after which all excerpts containing them are saved in filtered_df
import json

# Looping over all unique keywords and checking if they are present in the dataset
target_keywords = all_unique_keywords

def contains_target_keyword(cell):
    try:
        data = json.loads(cell)

        return any(
            item.get("keyword", "").lower() in target_keywords
            for item in data
        )

    except Exception:
        return False


filtered_df_final = combined_excerpts_needed[combined_excerpts_needed["Keywords With Timestamp"].apply(contains_target_keyword)]
filtered_df_final
````
